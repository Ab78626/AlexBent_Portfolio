**Preparing The data**

```python
df_prophet = pd.DataFrame(df.resample('M').size().reset_index())
```

```python
df_prophet.columns =['Date','Crime Count']
```

```python
df_prophet
```

**Rename The Columns**

```python
# ds and y
```

```python
df_prophet=df_prophet.rename(columns={'Date':'ds','Crime Count':'y'})
```

```python
df_prophet
```

**Train The Model**

```python
m = Prophet()
m.fit(df_prophet)
```

**Make The Predictions**

```python
pred=m.make_future_dataframe(periods=24,freq='M')
forcast = m.predict(pred)
```

```python
forcast
```

```python
m.plot(forcast,xlabel="Date",ylabel="Crime Rate")
```

**Save The Model**

```python
import json
from prophet.serialize import model_to_json, model_from_json

with open('prophet_model.json', 'w') as fout:
    json.dump(model_to_json(m), fout)  # Save model
```

```python
with open('prophet_model.json', 'r') as fin:
    m1 = model_from_json(json.load(fin))  # Load model
```

```python
x = int(input("Enter Numbers Months to forecast"))
```

```python
pred = m1.make_future_dataframe(periods=x ,freq = "M")
forcast = m1.predict(pred)
```

```python
import folium
from folium import Choropleth
```

```python
# Re-load the original data to get the 'Community Area' column
data1 = pd.read_csv("/content/drive/Crime_Datasets/Chicago_Crimes_2005_to_2007.csv",on_bad_lines='skip', low_memory=False)
data2 = pd.read_csv("/content/drive/Crime_Datasets/Chicago_Crimes_2008_to_2011.csv",on_bad_lines='skip', low_memory=False)
data3 = pd.read_csv("/content/drive/Crime_Datasets/Chicago_Crimes_2012_to_2017.csv",on_bad_lines='skip', engine='python')
data = pd.concat([data1,data2,data3])

# Fill missing values in 'Community Area' with a placeholder or drop rows
# For simplicity, let's fill with a placeholder like -1
data['Community Area'] = data['Community Area'].fillna(-1).astype(int)

# Group by Community Area and count crimes
crime_counts_by_area = data['Community Area'].value_counts().reset_index()
crime_counts_by_area.columns = ['community', 'Number of crimes']

# Filter out the placeholder if used
crime_counts_by_area = crime_counts_by_area[crime_counts_by_area['community'] != -1]

# Read the geojson file
read_map = pd.read_json('/content/drive/map/chicago_community_boundaries.json.xz')

# Create the map
mapchicagocrime = folium.Map(location=[41.8781, -87.6298], zoom_start=10, tiles='cartodbpositron')

# Create the Choropleth layer
Choropleth(
    geo_data=read_map,
    data=crime_counts_by_area,
    columns=['community', 'Number of crimes'],
    key_on='feature.properties.community',
    fill_color='YlOrRd',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Number of crimes by Community Area'
).add_to(mapchicagocrime)

# Add LayerControl to toggle the Choropleth layer
folium.LayerControl().add_to(mapchicagocrime)

# Display the map
display(mapchicagocrime)
```
